# DiffuseCraft — Text-to-Image on Google Colab

Run a free, open-source Stable Diffusion web app directly in your browser.

**Steps:**
1. `Runtime > Change runtime type > T4 GPU` (free tier works)
2. `Runtime > Run all`
3. Scroll to the bottom cell — a public `https://xxxx.gradio.live` URL will be printed. Open it in any browser.

Repo: https://github.com/shakib79ai/DiffuseCraft

## 3. Install dependencies

We install from `requirements-colab.txt`, which lists `torch` and
`torchvision` **together** alongside the rest. That pairing matters: if only
one of them gets touched by pip (e.g. torch silently upgraded as a
transitive dependency of diffusers/transformers, while the old torchvision
is left in place), the two become ABI-incompatible and imports fail with
`RuntimeError: operator torchvision::nms does not exist`. Listing both
explicitly forces pip to resolve them together in the same solve, so they
stay a matched pair. We also skip a separate `xformers` install — forcing a
specific CUDA wheel for it causes the same kind of mismatch, and `app.py`
already falls back gracefully to PyTorch 2.x's built-in scaled-dot-product
attention.

```
!pip install -q -r requirements-colab.txt
```

## 4. Launch the app

This starts the Gradio server with `share=True`, which prints a public
`*.gradio.live` URL. That link works from any browser (desktop or mobile)
and stays active for as long as this cell keeps running.

```
!python app.py
```

## Notes

- First run downloads the chosen model (~2-5 GB) from the Hugging Face Hub and caches it for the rest of the session.
- Free Colab GPUs (T4) comfortably run Stable Diffusion 1.5 / 2.1 at 512x512 in float16.
- The `gradio.live` link expires when the Colab runtime disconnects or after ~72 hours, whichever is first.
- To stop the server, interrupt the cell (`Runtime > Interrupt execution`).
